```python
# Analyze missingness

def missing_summary(df, vars=None):
    """ 
    Summarize missingess rate and count for each variable in the df
    If 'vars' is None, summarizes all variables in the df
    """


    # Define which cols to use
    cols = vars if vars is not None else df.columns

    summary = pd.DataFrame({
        "Count missing": df[cols].isna().sum(),
        "Percent missing": (df[cols].isna().mean()*100).round(2),
        "Number remaining": df[cols].notna().sum()
    })

    summary = summary.sort_values("Percent missing", ascending=False)

    return None
```

### Descriptive Statistics

```python
def summarize_cont_columns(df, cols):
    # Returns a table of the continuous variables, their mean,
    # standard deviation, median, IQR
    # Age, BMI, HbA1c, hsCRP, jetlag

    desc = df[cols].describe().T

    return desc


def summarize_cat_columns(df, cols):
    # Returns a table of the categorical variables, their categories,
    # the count, and percentage
    # Sex, race, smoking, PA?

    rows = []
    for col in cols:
        counts = df[col].value_counts()
        pct = df[col].value_counts(normaize=True) * 100
        rows.append(pd.Dataframe({'Variable': col, "Category": counts.index, "Count":counts.values, "pct": pct.values.round(1)}))

    return pd.concat(rows, ignore_index=True)
```

### Distribution Plots

```python
def plot_dist(df, column):
    # For each continuous variable, plot the distribution
    df[column].plot(kind="hist", bins=30, edgecolor="black")
    plt.title(column)
    plt.xlabel(column)
    plt.show()

    return  

def plot_counts(df, column):
    # plot the counts for each categorical variable
    df[column].value_counts().plot(kind="bar", edgecolor="black")
    plt.title(column)
    plt.xlabel(column)
    plt.ylabel("count")
    plt.show()

    return None
```

### Outlier detection

```python
def identify_outs(df, column):
    # Use boxplots and IQR

    # Visualize the variable
    df[column].plot(kind="box")
    plt.title(column)
    plt.show()

    # Use IQR to find outliers
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3-q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outs = df[(df[column] < lower) | (df[column] > upper)]

    # List how may
    print(f"{column}: {len(outs)} outliers (bounds:{lower:.2f} to {upper:.2f})")


    return outs
```

```python
# For the categorical varaibles defined earlier, find outliers

for col in cont_vars:
    outs = identify_outs(df,col)
```

```python
def plot_correlation_heatmap(df, variables):
    """
    Plot a correlation heatmap for selected continuous variables 
    
    Note: Spearman vs pearson? Spearman would work for hs-CRP since
    it frequently has a skewed distribution

    Works on continuous variables and categorical encoded as numeric
    """
    method = 'pearson'

    corr = df[variables].corr(method=method)
    plt.figure(figsize=(10,8))
    sns.heatmap(
        corr,
        annot=True,
        cmap="coolwarm",
        square=True,
    )

    plt.title(f"{method.capitalize()} Correlation Matrix")
    plt.tight_layout()
    plt.show()
```

```python
plot_correlation_heatmap(df, df.columns)
```

```python
def plot_pairplot(df, variables):
    """
    Generates a scatter plot matrix (pairplot) with KDE diagonals
    """

    sns.pairplot(
        df[variables],
        diag_kind="kde",
        corner = True
    )

    plt.show()
```

```python
plot_pairplot(df, df.columns)
```

```python
def plot_depression_by_category(df, category, outcome="DPQ_total"):
    """
    Plot the distribution of the DPQ-9 scores across levels of a
    categorical variable
    """
    
    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x=category,
        y=outcome
    )

    plt.title(f"{outcome} by {category}")
    plt.tight_layout()
    plt.show()
```

```python
# Visaulize for all categorical variables used

plot_depression_by_category(df, "sex")
plot_depression_by_category(df, "smoking_status")
```

```python
def plot_scatter(df, x, y):
    """
    Plots a scatter plot with regression line and 95% CI
    """

    plt.figure(figsize=(6,5))
    sns.regplot(
        data=df,
        x=x,
        y=y,
        scatter_kws={"alpha":0.4}, # make points somewhat opaque
        line_kws={"color":"red"} # 
    )

    plt.title(f"{y} vs {x}")
    plt.tight_layout()
    plt.show()
```

```python
plot_scatter(df, "social_jetlag", "DPQ9_total")
plot_scatter(df, "log_hscrp", "DPQ9_total")
plot_scatter(df, "HbA1c", "DPQ9_total")
plot_scatter(df, "sleep_duration", "DPQ9_total")
plot_scatter(df, "age", "DPQ9_total")
```
